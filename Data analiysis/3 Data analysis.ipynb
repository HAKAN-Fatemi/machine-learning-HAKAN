## استانداردسازی کامل داده‌ها (Standardization / Scaling)
داده‌های عددی را با استاندارد می‌کنیم (میانگین = 0، واریانس = 1).

**StandardScaler** : با استفاده از

اگر دیتافریم وجود نداشت، دیتا را می‌سازیم تا سلول مستقل اجرا شود


```python
# اطمینان از وجود df؛ در صورت نبودن، دیتافریم نمونه را می‌سازیم
try:
    df
except NameError:
    import pandas as pd
    data = {
        "name": ["Ali","Sara","Reza","Mina","Omid","Neda","Hasan","Maryam","Farid","Zahra"],
        "major": ["Math","Physics","CS","CS","Math","Physics","Math","CS","Physics","Math"],
        "math": [18,15,12,19,20,14,16,11,13,10],
        "physics": [17,14,15,13,18,12,19,12,16,14],
        "english": [16,18,14,12,17,15,13,19,11,18]}
    
    df = pd.DataFrame(data)
    df["avg"] = df[["math","physics","english"]].mean(axis=1)
```

```python
# استانداردسازی
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numeric_cols = ["math","physics","english","avg"]
df_std = df.copy()
df_std[numeric_cols] = scaler.fit_transform(df_std[numeric_cols])

# نمایش
print("نمونه‌ای از داده‌های استاندارد شده:")
df_std
```

## نمودار دایره‌ای (Pie Chart) — همه پارامترهای مهم
پارامترها:
- `labels` : برچسب‌ها
- `colors` : رنگ‌ها (لیست رنگ‌ها یا کدهای هگز)
- `autopct` : نمایش درصد داخل پِی (مثلاً "%1.1f%%")
- `startangle` : زاویه شروع (درجه)
- `shadow` : سایه
- `explode` : جدا کردن یک یا چند تکه برای تاکید
- `wedgeprops` : دیکشنری برای ویژگی‌های مربوط به هر بخش (مثل edgecolor, linewidth)


```python
majors_count = df["major"].value_counts()
labels = majors_count.index.tolist()
sizes = majors_count.values.tolist()

plt.figure(figsize=(5,5))

explode = [0.03]*len(labels)

plt.pie(
    sizes,
    labels=labels,
    colors=["#f11010","#0c75de","#22e922"][:len(labels)],
    autopct="%1.1f%%",
    startangle=140,
    shadow=True,
    explode=explode,
    wedgeprops={"edgecolor":"black","linewidth":1.2}
)

plt.title("Distribution of Majors (Pie Chart)")
plt.axis("equal")  # برای دایره‌ای بودن دقیق
plt.show()
```

## نمودار میله‌ای (Bar Chart)

#### width, color, edgecolor, linewidth, align : پارامترهای 

مثال: میانگین معدل بر اساس رشته


```python
avg_scores = df.groupby("major")["avg"].mean()
x = np.arange(len(avg_scores))

plt.figure(figsize=(8,6))
bars = plt.bar(
    x,
    avg_scores.values,
    width=0.5,                   # ضخامت میله
    color=["red","blue","yellow"][:len(avg_scores)],  # رنگ هر میله
    edgecolor="black",           # رنگ لبه میله
    linewidth=1.5,               # ضخامت خط لبه
    align="center"               # هم‌تراز کردن (center/edge)
)

# برچسب محور x
plt.xticks(x, avg_scores.index)
plt.xlabel("Major")
plt.ylabel("Average Score")
plt.title("Average Scores by Major — Bar Chart (with full params)")

# اضافه کردن مقادیر عددی بالای هر میله
for rect in bars:
    height = rect.get_height()
    plt.text(rect.get_x() + rect.get_width()/2, height + 0.1, f"{height:.2f}", ha="center", va="bottom")

plt.ylim(0, max(avg_scores.values)*1.2)
plt.show()
```

## Heatmap (نقشه همبستگی)

annot, fmt, linewidths, linecolor, cmap, cbar : پارامترهای

ما هم نقشهٔ همبستگی برای دادهٔ اصلی و هم برای دادهٔ استاندارد شده رسم می‌کنیم تا تفاوت را ببینیم


```python
import seaborn as sns

# همبستگی برای داده‌ی اصلی
plt.figure(figsize=(7,5))
corr_original = df[["math","physics","english","avg"]].corr()
sns.heatmap(
    corr_original,
    annot=True,
    fmt=".2f",
    linewidths=0.8,
    linecolor="black",
    cmap="coolwarm",
    cbar=True
)
plt.title("Correlation Heatmap — Original Data")
plt.show()

# همبستگی برای داده‌ی استاندارد شده
plt.figure(figsize=(7,5))
corr_std = df_std[["math","physics","english","avg"]].corr()
sns.heatmap(
    corr_std,
    annot=True,
    fmt=".2f",
    linewidths=0.8,
    linecolor="black",
    cmap="vlag",
    cbar=True
)
plt.title("Correlation Heatmap — Standardized Data")
plt.show()
```
