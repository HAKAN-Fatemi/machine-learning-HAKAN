### میانگین

```python
print((99+86+87+88+111+86+103+87+94+78+77+85+86) / 13)
print(sum(x) / len(x))
```

### میانه
"عدد وسط"

مهم است که اعداد قبل از یافتن میانه، مرتب شده باشند.

```python
np.median(x)
```

اگر دو عدد در وسط وجود دارد، مجموع آن اعداد را بر دو تقسیم کنید

```python
x = [99,86,87,88,86,103,87,94,78,77,85,86]

print(x)

x.sort()
print(x)

np.median(x)
```

### Mode

مقداری است که بیشترین تعداد بار ظاهر می‌شود

```python
from scipy import stats

s = [99,86,87,88,111,86,103,87,94,78,77,85,86]

x = stats.mode(s)

print(x)
```

# انحراف معیار


انحراف معیار عددی است که میزان پراکندگی مقادیر را توصیف می‌کند.

انحراف معیار پایین به این معنی است که بیشتر اعداد نزدیک به مقدار میانگین (میانگین) هستند.

انحراف معیار بالا به این معنی است که مقادیر در محدوده وسیع‌تری پخش شده‌اند

```python
s = [86,87,88,86,87,85,86]

x = np.std(s)

print(x)
```

به این معنی که بیشتر مقادیر در محدوده ۰.۹ از مقدار میانگین، که ۸۶.۴ است، قرار دارند

# واریانس

واریانس عدد دیگری است که میزان پراکندگی مقادیر را نشان می‌دهد

در واقع، اگر جذر واریانس را بگیرید، انحراف معیار را به دست می‌آورید

یا برعکس، اگر انحراف معیار را در خودش ضرب کنید، واریانس به دست می‌آید

## فرمول
۱. میانگین را بیابید

۲. برای هر مقدار: اختلاف را از میانگین پیدا کنید

۳. برای هر تفاضل: مقدار مربع را بیابید

۴. واریانس، میانگین تعداد این مربعات تفاوت‌ها است

(32+111+138+28+59+77+97) / 7 = 77.4

-----------------

32 - 77.4 = -45.4

111 - 77.4 =  33.6

138 - 77.4 =  60.6

28 - 77.4 = -49.4

59 - 77.4 = -18.4

77 - 77.4 = - 0.4

97 - 77.4 =  19.6

-----------------

(-45.4)2 = 2061.16

(33.6)2 = 1128.96

(60.6)2 = 3672.36

(-49.4)2 = 2440.36

(-18.4)2 =  338.56

(- 0.4)2 =    0.16

(19.6)2 =  384.16

-----------------

(2061.16+1128.96+3672.36+2440.36+338.56+0.16+384.16) / 7 = 1432.2

```python
s = [32,111,138,28,59,77,97]

x = np.var(s)

print(x)
```

# صدک‌ها 


صدک‌ها در آمار برای ارائه عددی استفاده می‌شوند که مقداری را توصیف می‌کند که درصد مشخصی از مقادیر از آن کمتر هستند

```python
ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]
```

صدک ۷۵ چقدر است؟ جواب ۴۳ است، به این معنی که ۷۵٪ از افراد ۴۳ سال یا کمتر سن دارند

```python
ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]

x = np.percentile(ages, 75)

print(x)
```
