برای ایجاد مجموعه داده‌های بزرگ برای آزمایش،
ما از ماژول‌های پایتون استفاده می‌کنیم که دارای تعدادی روش برای ایجاد مجموعه داده‌های تصادفی، با هر اندازه‌ای، است

```python
# یک آرایه شامل ۲۵۰ عدد اعشاری تصادفی بین ۰ تا ۵ ایجاد می‌کند

import numpy as np

x = np.random.uniform(0.0, 5.0, 250)

print(x)
```

برای تجسم مجموعه داده‌ها، می‌توانیم یک هیستوگرام با داده‌های جمع‌آوری‌شده رسم کنیم

```python
import matplotlib.pyplot as plt

x = np.random.uniform(0.0, 5.0, 250)

plt.hist(x, 5)
plt.show()
```


نوار اول نشان می‌دهد که چه تعداد از مقادیر موجود در آرایه بین ۰ و ۱ هستند

نوار دوم نشان دهنده تعداد مقادیر بین ۱ و ۲ است

و غیره

که این نتیجه را به ما می‌دهد

- ۵۲ مقدار بین ۰ و ۱ هستند
- ۴۸ مقدار بین ۱ و ۲ هستند
- ۴۹ مقدار بین ۲ و ۳ هستند
- ۵۱ مقدار بین ۳ و ۴ هستند
- ۵۰ مقدار بین ۴ و ۵ هستند
______

# مثال
یک آرایه با ۱۰۰۰۰۰ عدد تصادفی ایجاد کنید و آنها را با استفاده از یک هیستوگرام با ۱۰۰ میله نمایش دهید

```python
x = np.random.uniform(0.0, 5.0, 100000)

plt.hist(x, 100)
plt.show()
```

# Normal Data Distribution / توزیع نرمال داده‌ها

در این بخش یاد خواهیم گرفت که چگونه آرایه‌ای ایجاد کنیم که مقادیر آن حول یک مقدار معین متمرکز شده باشند

```python
x = np.random.normal(5.0, 1, 100000)

plt.hist(x, 100)
plt.show()
```

ما مشخص می‌کنیم که مقدار میانگین ۵.۰ و انحراف معیار ۱.۰ است

به این معنی که مقادیر باید حول و حوش عدد ۵.۰ متمرکز شوند و به ندرت دورتر از ۱.۰ از میانگین قرار گیرند

و همانطور که از هیستوگرام می‌بینید، بیشتر مقادیر بین ۴.۰ و ۶.۰ هستند و بالاترین مقدار تقریباً ۵.۰ است

```python
x = np.random.normal(5.0, 1.0, 1000)
y = np.random.normal(10.0, 2.0, 1000)

plt.scatter(x, y)
```

# Scale / مقیاس

وقتی داده‌های شما مقادیر و حتی واحدهای اندازه‌گیری متفاوتی دارند، مقایسه آنها می‌تواند دشوار باشد

### روش استانداردسازی از این فرمول استفاده می‌کند
z = (x - u) / s 
===
z مقدار جدید

u میانگین 

s انحراف معیار

```python
import pandas
from sklearn import linear_model

from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
```

```python
import requests

url = "https://raw.githubusercontent.com/HAKAN-Fatemi/machine-learning-HAKAN/main/file_csv/data.csv"
response = requests.get(url)

with open("data.csv", "wb") as f:
    f.write(response.content)
```

```python
df = pandas.read_csv("data.csv")

X = df[['Weight', 'Volume', "CO2"]]
df
```

```python
scaledX = scale.fit_transform(X)

print(scaledX)
```

یا میتوان به شکل زیر نوشت

```python
scaledX = scale.fit(X)
scaledX = scale.transform(X)

print(scaledX)
```
